```python
import torch
import transformers
import peft
import bitsandbytes
import sentence_transformers
import faiss

print("All required libraries imported successfully!")
print("PyTorch:", torch.__version__)
print("Transformers:", transformers.__version__)
print("PEFT:", peft.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
import torch
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig
)

MODEL_NAME = "Qwen/Qwen2.5-1.5B-Instruct"

# Configure 4-bit quantisation
quant_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True
)

print("Loading tokenizer...")

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

print("Loading model...")

base_model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=quant_config,
    device_map="auto"
)

base_model.eval()

print("\nModel loaded successfully!")
print("Model:", MODEL_NAME)
print("GPU:", torch.cuda.get_device_name(0))
```

```python
def generate_answer(question, max_new_tokens=200):

    messages = [
        {
            "role": "system",
            "content": (
                "You are a medical question-answering assistant. "
                "Provide clear, cautious, evidence-aware information. "
                "Do not invent medical facts. Encourage professional "
                "medical care when appropriate."
            )
        },
        {
            "role": "user",
            "content": question
        }
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to("cuda")

    with torch.no_grad():
        outputs = base_model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    generated_tokens = outputs[0][inputs["input_ids"].shape[1]:]

    answer = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    )

    return answer.strip()
```

```python
test_question = (
    "What are some common symptoms of iron deficiency "
    "and when should someone seek medical advice?"
)

baseline_answer = generate_answer(test_question)

print("QUESTION:")
print(test_question)

print("\nBASELINE MODEL ANSWER:")
print(baseline_answer)
```

```python
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np

# Load a small embedding model
embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

# Prepare documents from the training sample
rag_documents = []

for index, row in prototype_train.iterrows():
    document = (
        f"Question: {row['Question']}\n"
        f"Answer: {row['Answer']}"
    )
    rag_documents.append(document)

print("Number of RAG documents:", len(rag_documents))
print("Example document:")
print(rag_documents[0][:500])
```

```python
# Convert documents into numerical embeddings
embeddings = embedding_model.encode(
    rag_documents,
    convert_to_numpy=True,
    show_progress_bar=True
)

# Convert embeddings to float32
embeddings = embeddings.astype("float32")

# Normalize embeddings for similarity search
faiss.normalize_L2(embeddings)

# Create the FAISS index
embedding_dimension = embeddings.shape[1]

rag_index = faiss.IndexFlatIP(embedding_dimension)

# Add document embeddings to the index
rag_index.add(embeddings)

print("FAISS index created successfully!")
print("Number of indexed documents:", rag_index.ntotal)
print("Embedding dimensions:", embedding_dimension)
```

```python
def retrieve_documents(question, top_k=3):

    # Convert the question into an embedding
    question_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True
    ).astype("float32")

    # Normalize the question embedding
    faiss.normalize_L2(question_embedding)

    # Search the FAISS index
    scores, indices = rag_index.search(
        question_embedding,
        top_k
    )

    # Collect the retrieved documents
    retrieved_documents = []

    for score, index in zip(scores[0], indices[0]):
        if index != -1:
            retrieved_documents.append({
                "document": rag_documents[index],
                "similarity_score": float(score)
            })

    return retrieved_documents


# Test the retrieval function
test_question = "What are common symptoms of iron deficiency?"

retrieved = retrieve_documents(test_question)

print("TEST QUESTION:")
print(test_question)

print("\nRETRIEVED DOCUMENTS:")

for i, item in enumerate(retrieved, 1):
    print(f"\nDocument {i}")
    print("Similarity score:", round(item["similarity_score"], 3))
    print(item["document"][:500])
```

```python
def generate_rag_answer(question, retrieved_documents):

    context = "\n\n".join(
        [item["document"] for item in retrieved_documents]
    )

    messages = [
        {
            "role": "system",
            "content": (
                "You are a cautious medical question-answering assistant. "
                "Use the retrieved context as supporting information, "
                "but do not assume it is always correct. "
                "Do not invent medical facts. "
                "If information is uncertain or missing, say so. "
                "Encourage professional medical advice when appropriate."
            )
        },
        {
            "role": "user",
            "content": (
                f"Retrieved context:\n{context}\n\n"
                f"Question: {question}\n\n"
                "Provide a clear and cautious answer."
            )
        }
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=2048
    ).to("cuda")

    with torch.no_grad():
        outputs = base_model.generate(
            **inputs,
            max_new_tokens=200,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    generated_tokens = outputs[0][
        inputs["input_ids"].shape[1]:
    ]

    answer = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    )

    return answer.strip()


# Test the RAG system
rag_answer = generate_rag_answer(
    test_question,
    retrieved
)

print("QUESTION:")
print(test_question)

print("\nRAG MODEL ANSWER:")
print(rag_answer)
```

```python
# Convert training examples into Qwen conversational format

def format_lora_example(row):
    messages = [
        {
            "role": "system",
            "content": (
                "You are a cautious medical question-answering assistant. "
                "Provide clear and evidence-aware answers. "
                "Do not invent medical facts."
            )
        },
        {
            "role": "user",
            "content": row["Question"]
        },
        {
            "role": "assistant",
            "content": row["Answer"]
        }
    ]

    return tokenizer.apply_chat_template(
        messages,
        tokenize=False
    )


# Format training and validation data
train_texts = [
    format_lora_example(row)
    for _, row in prototype_train.iterrows()
]

validation_texts = [
    format_lora_example(row)
    for _, row in prototype_validation.iterrows()
]

print("Training examples formatted:", len(train_texts))
print("Validation examples formatted:", len(validation_texts))

print("\nExample formatted training record:\n")
print(train_texts[0][:1000])
```

```python
from peft import LoraConfig, prepare_model_for_kbit_training, get_peft_model

# Prepare the 4-bit model for LoRA training
base_model.gradient_checkpointing_enable()

base_model = prepare_model_for_kbit_training(base_model)

# Configure LoRA
lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    target_modules=[
        "q_proj",
        "k_proj",
        "v_proj",
        "o_proj"
    ],
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM"
)

# Attach LoRA adapters to the model
lora_model = get_peft_model(
    base_model,
    lora_config
)

# Display trainable parameters
lora_model.print_trainable_parameters()
```

```python
from datasets import Dataset

# Create Hugging Face datasets
train_dataset = Dataset.from_dict({
    "text": train_texts
})

validation_dataset = Dataset.from_dict({
    "text": validation_texts
})

# Tokenisation function
def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        truncation=True,
        max_length=512,
        padding="max_length"
    )

# Tokenise training and validation datasets
tokenized_train = train_dataset.map(
    tokenize_function,
    batched=True
)

tokenized_validation = validation_dataset.map(
    tokenize_function,
    batched=True
)

print("Training dataset:", len(tokenized_train))
print("Validation dataset:", len(tokenized_validation))
print("\nTokenised columns:", tokenized_train.column_names)
```

```python
from transformers import (
    DataCollatorForLanguageModeling,
    TrainingArguments,
    Trainer
)

# Remove the original text column
tokenized_train = tokenized_train.remove_columns(["text"])
tokenized_validation = tokenized_validation.remove_columns(["text"])

# Prepare the data collator
data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

# Configure training
training_args = TrainingArguments(
    output_dir=os.path.join(MODEL_DIR, "lora_prototype"),
    num_train_epochs=1,
    per_device_train_batch_size=1,
    per_device_eval_batch_size=1,
    gradient_accumulation_steps=4,
    learning_rate=2e-4,
    fp16=True,
    logging_steps=10,
    evaluation_strategy="epoch",
    save_strategy="epoch",
    save_total_limit=1,
    report_to="none"
)

print("LoRA training configuration created successfully!")
print("Epochs:", training_args.num_train_epochs)
print("Training batch size:", training_args.per_device_train_batch_size)
print("Learning rate:", training_args.learning_rate)
print("FP16 enabled:", training_args.fp16)
```

```python
trainer = Trainer(
    model=lora_model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_validation,
    data_collator=data_collator
)

print("LoRA Trainer created successfully!")
```

```python
# Start LoRA fine-tuning
training_result = trainer.train()

print("\nLoRA training completed successfully!")
print("Training loss:", training_result.training_loss)
print("Training steps:", training_result.global_step)
```
