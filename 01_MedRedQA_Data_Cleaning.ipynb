```python
import pandas as pd

file_path = "/content/drive/MyDrive/CSCU9Z7/medredqa_train.csv"

df = pd.read_csv(file_path)

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst five records:")
display(df.head())

print("\nDataset information:")
df.info()
```

```python
# 1. Check missing values
print("MISSING VALUES")
print(df.isnull().sum())

# 2. Check duplicate records
print("\nDUPLICATE RECORDS")
print("Exact duplicate rows:", df.duplicated().sum())

print("Duplicate question titles:", df["Title"].duplicated().sum())

# 3. Check response lengths
df["response_length"] = df["Response"].fillna("").str.len()

print("\nRESPONSE LENGTH STATISTICS")
print(df["response_length"].describe())

# 4. Check question lengths
df["question_length"] = (
    df["Title"].fillna("") + " " + df["Body"].fillna("")
).str.len()

print("\nQUESTION LENGTH STATISTICS")
print(df["question_length"].describe())

# 5. Check response scores
print("\nRESPONSE SCORE DISTRIBUTION")
print(df["Response Score"].value_counts().sort_index())
```

```python
import os

PROJECT_DIR = "/content/drive/MyDrive/CSCU9Z7"
CLEAN_DIR = os.path.join(PROJECT_DIR, "cleaned_datasets")

os.makedirs(CLEAN_DIR, exist_ok=True)

# Create a working copy
df_working = df.copy()

# Remove the automatically generated index column
if "Unnamed: 0" in df_working.columns:
    df_working = df_working.drop(columns=["Unnamed: 0"])

print("Working dataset shape:", df_working.shape)
print("Original dataset remains unchanged.")
```

```python
# Find records with repeated titles
duplicate_titles = df_working[
    df_working["Title"].duplicated(keep=False)
].sort_values("Title")

print("Records with repeated titles:", len(duplicate_titles))

display(
    duplicate_titles[
        ["Title", "Body", "Response"]
    ].head(10)
)
```

```python
import pandas as pd
import os

# Start from the original working copy
df_clean = df_working.copy()

# Remove diagnostic columns if present
diagnostic_columns = ["response_length", "question_length"]

df_clean = df_clean.drop(
    columns=[
        col for col in diagnostic_columns
        if col in df_clean.columns
    ]
)

# Remove records with missing responses
df_clean = df_clean.dropna(subset=["Response"]).copy()

# Remove responses that are empty or contain only whitespace
df_clean = df_clean[
    df_clean["Response"].astype(str).str.strip() != ""
].copy()

# Remove leading/trailing whitespace from text fields
for col in ["Title", "Body", "Response"]:
    df_clean[col] = df_clean[col].astype(str).str.strip()

# Combine title and body into one question field
df_clean["Question"] = (
    df_clean["Title"] + "\n" + df_clean["Body"]
)

# Keep the answer in a separate field
df_clean["Answer"] = df_clean["Response"]

print("Original records:", len(df_working))
print("Cleaned records:", len(df_clean))
print("Removed records:", len(df_working) - len(df_clean))

print("\nFinal columns:")
print(df_clean.columns.tolist())

display(df_clean[["Question", "Answer"]].head(3))
```

```python
CLEAN_DIR = "/content/drive/MyDrive/CSCU9Z7/cleaned_datasets"

os.makedirs(CLEAN_DIR, exist_ok=True)

output_path = os.path.join(
    CLEAN_DIR,
    "medredqa_train_prepared.csv"
)

df_clean.to_csv(output_path, index=False)

print("Saved successfully:")
print(output_path)
```
